```python
# Imports
import numpy as np
import pandas as pd

# Data Visualization
import seaborn as sns
import matplotlib.pyplot as plt
import colorcet as cc
%matplotlib inline
import ipywidgets

import warnings
warnings.filterwarnings("ignore")
```

```python
# Read CSV file & Display CSV file
df = pd.read_csv("data/nbateamstats_final.csv")
with pd.option_context('display.max_rows', None, 'display.max_columns', None,'display.precision', 3,):
    display(df)
```

```python
# Checks the data type and the missing values, there should be no missing values in the Data
df.info()
```

```python
# Makes a copy of dataset in order to change datatype of floats -> integers
# Ensures no alteration in the oridinal df
df_int = df.copy()
df_int[['fgp', 'ftp', 'tpp']] = df_int[['fgp', 'ftp', 'tpp']].astype(int)
display(df_int)
```

```python
# Dropping the unused columns in the data
df_int.drop(columns=['teamID', 'games', 'biggestLead', 'longestRun', 'points', 'plusMinus'], inplace=True)
df_int
```

```python
# Display full dataset
with pd.option_context('display.max_columns', None):
    display(df_int.describe().style.set_sticky(axis="index").set_sticky(axis="columns"))# Generates summary statistics
```

```python
with pd.option_context('display.max_columns', None):
    display(df_int)
```

```python
df_int.sort_values(by = ['conferenceRank'], inplace = True)
with pd.option_context('display.max_columns', None):
    display(df_int)
```

```python
# Correlation Matrix, Full-Sized
correlation_matrix = df_int.corr(numeric_only=True)

plt.figure(figsize = (17, 15))
sns.heatmap(correlation_matrix, annot = True, cmap = 'coolwarm', fmt = ".2f", linewidths = 0.5)
plt.title('Correlation Matrix')
plt.show()
```

```python
# Correlation Matrix on the important features that affect wins at home
numeric_df = df_int.select_dtypes(include='number')
df_int_winHome = numeric_df.corrwith(numeric_df['winHome']) # correlation to winHome
plt.figure(figsize = (5, 7))
sns.heatmap(df_int_winHome.to_frame(), annot = True, linewidths = 0.5, cmap = 'YlGnBu')
plt.title('Correlation Matrix between Team\'s Stats and Winning at Home')
plt.show()

# pointsOffTurnovers 0.3
# fgm 0.46
# tpm 0.29
# defReb 0.41
# assists 0.29
# steals 0.29
# fgm is the highest feature correlating with a team winning home games
# defReb has a higher correlation than steals in regards to defensive features 
```

```python
# --- East conf scatter + regression (keep non-numeric cols) ---
# Make sure required columns exist & coerce numerics used in the plot
for col in ['conferenceName', 'teamCode', 'winHome', 'defReb']:
    if col not in df_int.columns:
        raise KeyError(f"Column missing: {col}")

df_int['winHome'] = pd.to_numeric(df_int['winHome'], errors='coerce')
df_int['defReb']  = pd.to_numeric(df_int['defReb'],  errors='coerce')

# Filter from the FULL df (not numeric_df), case-insensitive
east_mask = df_int['conferenceName'].str.lower() == 'east'
df_east = df_int.loc[east_mask].copy().sort_values('winHome', ascending=False)

# Build a palette (try colorcet.glasbey, else fallback)
try:
    import colorcet as cc
    custom_palette = sns.color_palette(cc.glasbey, n_colors=df_east['teamCode'].nunique())
except Exception:
    custom_palette = sns.color_palette("tab20", n_colors=df_east['teamCode'].nunique())

# Plot
sns.lmplot(
    data=df_east,
    x='winHome', y='defReb',
    hue='teamCode',
    scatter=True,
    scatter_kws={'s': 200},
    palette=custom_palette,
    line_kws={'color': 'orange'},
    legend=False,
    fit_reg=True,
    height=6, aspect=1
)
plt.title("Defensive Rebounds for Eastern Conf. Teams (sorted by Home Wins)")
plt.xlabel("Wins Home")
plt.ylabel("Defensive Rebounds")
plt.legend(title='Team', loc='upper right', bbox_to_anchor=(1.35, 1.0), ncol=1)
plt.show()
```

```python
# Correlation matrix on important features to win away

df_int_winAway = df_int.corrwith(df_int['winAway'])

plt.figure(figsize=(5, 7))
sns.heatmap(df_int_winAway.to_frame(), annot=True, linewidths=0.5, cmap='YlGnBu')
plt.title('Correlation Matrix between Team\'s Stats and Winning Away')
plt.show()
    
```
